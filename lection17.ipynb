# XML

```python
import xml.etree.ElementTree as ET

path = r"e:/work/devproject_hg/python/course/tests/data/country_data.xml"
t = ET.parse(path)
print(t.getroot(), t.getroot().tag, t.getroot().attrib)

for country in t.getroot():
    print(country.tag, country.attrib.get("name"))
    for i in country:
        if i.text is not None:
            print(i.text)
```

## Creating XML document

```python
import xml.etree.ElementTree as ET

root = ET.Element("names")
ET.SubElement(root, "name", {"value" : "Elizabeth"})
ET.SubElement(root, "name", {"value" : "Thomas"})
ET.SubElement(root, "name", {"value" : "Jeremy"})
ET.dump(root)

tree = ET.ElementTree(root)
tree.write("e:/work/devproject_hg/python/course/tests/data/lection18_names.xml", encoding="utf-8",)
```

```python
import json

names = [
    { "name" : "Elizabeth"},
    { "name" : "Thomas"},
    { "name" : "Jeremy"}
]

print(json.dumps(names, indent=4))

path = r"e:/work/devproject_hg/python/course/tests/data/lection18_names.json"
with open(path, "w") as f:
    json.dump(names, f, indent=4)
```
